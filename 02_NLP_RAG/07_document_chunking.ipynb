```python
words = policy_text.split()
chunks = []
current_chunk = []
current_len = 0

for word in words:
    word_len = len(word) + 1  # +1 for the space
    if current_len + word_len > chunk_size:
        chunks.append(" ".join(current_chunk))
        current_chunk = [word]
        current_len = word_len
    else:
        current_chunk.append(word)
        current_len += word_len

if current_chunk:
    chunks.append(" ".join(current_chunk))
```

```python
from langchain_text_splitters import CharacterTextSplitter,RecursiveCharacterTextSplitter


splitter = CharacterTextSplitter(
    separator=".",
    chunk_size=30,
    chunk_overlap=10
)


# separator = "\n\n"   # split by paragraph
# separator = "\n"     # split by line
# separator = " "      # split by word space
# separator = "."      # split by sentence end
# separator = ","      # split by comma
# separator = ""       # split by character

chunks = splitter.split_text(policy_text)
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

pdf_path = "bajaj_finance_policy_prose_v1.pdf"
loader = PyPDFLoader(pdf_path)
raw_docs = loader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=256,
    chunk_overlap=50,
    separators=["\n\n", "\n", ". ", " ", ""],
)

chunks = splitter.split_documents(raw_docs)
```
