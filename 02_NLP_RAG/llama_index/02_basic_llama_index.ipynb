```python
from llama_index.core import Settings
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
from llama_index.llms.openai import OpenAI
from llama_index.core import SimpleDirectoryReader
from llama_index.core import VectorStoreIndex # we will discuss this in future -->
from llama_index.core.response.pprint_utils import pprint_response # pretty print response

Settings.embed_model = HuggingFaceEmbedding(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    
)

Settings.llm = OpenAI(model="gpt-4o-mini")
```

```python
documents = SimpleDirectoryReader('data_test').load_data()
index = VectorStoreIndex(documents)
query_engine = index.as_query_engine()
```

```python
response = query_engine.query("please summarize the entire documents")
pprint_response(response,show_source=True)
```
