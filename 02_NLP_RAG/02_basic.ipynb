```python
DOCS = [
    "Gold loan foreclosure attracts 2% penalty if closed within 6 months of disbursal. No penalty after 6 months.",
    "Personal loan EMI is auto-debited on the 5th of every month. Insufficient funds attract a late fee of Rs 1000.",
    "Customers with CIBIL score below 650 are ineligible for a new personal loan or top-up loan.",
    "Loan renewal is available for Tier 1 and Tier 2 city customers with repayment score above 700 and no defaults in last 12 months.",
    "Gold loan renewal maximum is Rs 1,00,000 for salaried customers and Rs 50,000 for self-employed customers.",
    "Personal loan prepayment is allowed after 12 months of disbursal with a prepayment fee of 2% of the outstanding principal.",
    "Business loan prepayemnt is allowed after 15 month"
]
```

```python
def tokenize(text):
    return text.lower().split()



all_word = set()
for i in DOCS:
    word = tokenize(i)
    for w in word:
        all_word.add(w)

```

```python
VOCAB = list(all_word)
```

```python
# len(VOCAB)
```

```python
def bow_vector(chunk):
    token = tokenize(chunk)
    vector = []
    for w in VOCAB:
        count = token.count(w)
        vector.append(count)
    return vector
```

```python
# v1 = bow_vector("Gold loan foreclosure emi attracts two percent penalty if  emi closed within six months")
```

```python
DOCS_VECTOR = []

for i in DOCS:
    vector = bow_vector(i)
    DOCS_VECTOR.append(vector)
```

```python
# pip install scikit-learn
```

```python
from sklearn.metrics.pairwise import cosine_similarity
def search(query,top_k = 3,thresold=0.3):
    query_vector = bow_vector(query)
    similarities = []
    for doc_vector in DOCS_VECTOR:
        sim = cosine_similarity([query_vector], [doc_vector])
        similarities.append(sim)
    # Get top_k most similar documents where similarity > threshold
    top_k_indices = sorted(range(len(similarities)), key=lambda i: similarities[i], reverse=True)[:top_k]
    top_k_indices = [i for i in top_k_indices if similarities[i] > thresold]

    context = [DOCS[i] for i in top_k_indices]
    sorted_similarities = sorted(similarities, reverse=True)[:top_k]
    return context, sorted_similarities
```

```python
search("Gold loan foreclosure emi attracts two percent penalty if  emi closed within six months")
```

```python
import math
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import TfidfVectorizer


tfidf = TfidfVectorizer()


DOCS = [
    "Gold loan foreclosure attracts 2% penalty if closed within 6 months of disbursal. No penalty after 6 months.",
    "Personal loan EMI is auto-debited on the 5th of every month. Insufficient funds attract a late fee of Rs 1000.",
    "Customers with CIBIL score below 650 are ineligible for a new personal loan or top-up loan.",
    "Loan renewal is available for Tier 1 and Tier 2 city customers with repayment score above 700 and no defaults in last 12 months.",
    "Gold loan renewal maximum is Rs 1,00,000 for salaried customers and Rs 50,000 for self-employed customers.",
]

doc_mat = tfidf.fit_transform(DOCS)
```

```python
def tfidf_search(query,top_k=3):
    query_vector = tfidf.transform([query])
    sims = cosine_similarity(doc_mat,query_vector)


    result = []
    for i in range(len(sims)):
        result.append((sims[i],DOCS[i]))
    
    return sorted(result,reverse=True)[:top_k]



tfidf_search("Hi how  you")  
```
